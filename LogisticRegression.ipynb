"Importar librerias"

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score
from sklearn.datasets import make_classification
```

```python
# 1. tenemos una base de datos con 500 muestras
# 2. tenemos dos features
# 3. no hay features redundantes
# 4. numero de features informativos
X, y = make_classification(
    n_samples=500,
    n_features=2,
    n_redundant=0,
    n_informative=2,
    n_clusters_per_class=1,
    random_state=42
)
print(X)
print(y)
```

```python
#Dividimos daros para entrenamiento y pruebas
X_train, X_test, y_train, y_test = train_test_split(
X,
y,
test_size=0.3,
random_state=42
 )
```

```python
#Escalamos datos para procesar (preprocesamiento)
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
print(X_train)
print(X_test)
```

```python
#Generar nuestro modelo y lo alimentamos
model= LogisticRegression()
model.fit(X_train, y_train) #Para aliementar el modelo con X_train & y_train
```

```python
#Generamos prediccion
y_pred = model.predict(X_test)
print(y_pred)
```

```python
print("accuracy", accuracy_score(y_test, y_pred))
print("Confusion matrix:", confusion_matrix(y_test, y_pred))
print("Classification report:", classification_report(y_test, y_pred))
```

```python
def plot_decision_boundary(X, y, model):
  x_min, x_max = X[:,0].min() - 1, X[:,0].max() + 1
  y_min, y_max = X[:,1].min() - 1, X[:,1].max() + 1
  xx, yy = np.meshgrid(
      np.linspace(x_min, x_max, 200),
      np.linspace(y_min, y_max, 200)
  )
  Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
  Z = Z.reshape(xx.shape)
  plt.contourf(xx, yy, Z, alpha=0.3)
  plt.scatter(X[:,0], X[:,1], c=y, edgecolors="k", s=50)
  plt.xlabel("Feature 1")
  plt.ylabel("Feature 2")
  plt.title("Logistic Regression Decision Boundary")
  plt.show()
```

```python
plot_decision_boundary(X_train, y_train, model)
```

##Pipepline

##Importamos las librerias necesarias para hacer el PIPEPLINE

```python
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
```

#1. Cargamos los datos

```python
X, y = make_classification(n_samples=500, n_features=2, n_redundant=0,
                           n_informative=2, n_clusters_per_class=1, random_state=42)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

##2. Definicion de pipeline
Se agrupa el escalado y el modelo en un solo objeto llamado 'pipe'

```python
pipe = Pipeline([
    ('scaler', StandardScaler()),       #Escalar
    ('clf', LogisticRegression())      #Clasificar
])
```

##3. Entrenamiento
El pipe escala los datos automaticamente y luego entrena el modelo.

```python
pipe.fit(X_train, y_train)
```

4. Prediccion y evaluacion

```python
y_pred = pipe.predict(X_test)
print(f"Accuracy con Pipeline: {accuracy_score(y_test, y_pred):.4f}")
```

##5. Función de grafica usando el pipeline


```python
def plot_decision_boundary(X, y, pipeline_model):
    x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
    xx, yy = np.meshgrid(np.linspace(x_min, x_max, 200), np.linspace(y_min, y_max, 200))
    Z = pipeline_model.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)

    plt.figure(figsize=(10, 6))
    plt.contourf(xx, yy, Z, alpha=0.3, cmap='RdYlBu')
    plt.scatter(X[:, 0], X[:, 1], c=y, edgecolors="k", s=50, cmap='RdYlBu')
    plt.title("Frontera de Decisión (Usando Sklearn Pipeline)")
    plt.show()
```

```python
plot_decision_boundary(X_train, y_train, pipe)
```
